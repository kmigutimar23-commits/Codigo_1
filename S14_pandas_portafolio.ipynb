## 00 · ¿Qué es pandas y para qué se usa?

> **Nota** — Objetivo: entender qué problema resuelve pandas, y por qué lo aprendido en NumPy no es suficiente.

Un portafolio real no es solo números: tiene tickers (texto), cantidades (enteros), precios (decimales) y, más adelante, fechas. NumPy asume que todo el array es del mismo tipo — por eso necesitamos algo que junte columnas de distinto tipo en una sola tabla, con nombres. Eso es exactamente lo que resuelve **pandas**.

pandas es una librería que permite trabajar con datos organizados en tablas: leerlos, explorarlos, limpiarlos, filtrarlos y resumirlos. Está construida sobre NumPy — por dentro, cada columna de una tabla de pandas es un array.

> 💡 **Idea clave** — NumPy nos ayuda a calcular rápidamente sobre muchos números. Por otro lado, pandas nos permite organizar datos de distinto tipo en tablas que podemos leer, filtrar y consultar. Es importante destacar que no compiten, sino que pandas utiliza NumPy internamente.

## 01 · La estructura: Series y DataFrame

> **Nota** — Objetivo: distinguir las dos estructuras de datos principales que ofrece pandas.

pandas nos ofrece dos tipos de estructuras:

- **Series** — un arreglo unidimensional etiquetado, capaz de contener datos de cualquier tipo. Es una sola columna, con índice.
- **DataFrame** — una estructura de dos dimensiones que guarda datos como una tabla, con filas y columnas. Cada columna de un DataFrame es, de hecho, una Series.

````python
import numpy as np
import pandas as pd

precios = pd.Series([195, 385, 185, 165])
print(precios)
print(type(precios))
````

````python
portafolio = pd.DataFrame({
    "ticker": ["AAPL", "MSFT", "AMZN", "GOOGL"],
    "cantidad": [10, 8, 12, 6],
    "precio": [195, 385, 185, 165]
})
print(portafolio)
print(type(portafolio))
````

## 05 · Operaciones más usadas

> **Nota** — Objetivo: Agregar columnas calculadas y resumir un DataFrame de manera eficiente, sin necesidad de ciclos explícitos.

Agregar una columna nueva es tan simple como asignarla — pandas alinea los valores automáticamente.

**✏️ Ahora, completemos el código**

````python
portafolio["valor_actual"] = portafolio["cantidad"] * portafolio["precio_actual"]
portafolio["ganancia"] = portafolio["valor_actual"] - (portafolio["cantidad"] * portafolio["precio_compra"])

print(portafolio["ganancia"].mean())   # promedio
print(portafolio["ganancia"].std())   # desviación estándar
portafolio.sort_values(by="ganancia")
````

| Operación | Lo que hace |
|---|---|
| `df["col"] = ...` | crea o reemplaza una columna |
| `df["col"].mean() / .std()` | estadísticas sobre una columna |
| `df.sort_values(by="col")` | ordena las filas según una columna |
| `df[condición]` | filtra filas |

## 06 · Todo junto: el portafolio como DataFrame

> **Nota** — Objetivo: Integraremos la creación, el cálculo de columnas, el filtrado y la ordenación en un flujo unificado.

````python
portafolio = pd.DataFrame({
    "ticker": ["AAPL", "MSFT", "AMZN", "GOOGL"],
    "cantidad": [10, 8, 12, 6],
    "precio_compra": [180, 400, 170, 150],
    "precio_actual": [195, 385, 185, 165]
})

portafolio["valor_actual"] = portafolio["cantidad"] * portafolio["precio_actual"]
portafolio["ganancia"] = portafolio["valor_actual"] - (portafolio["cantidad"] * portafolio["precio_compra"])
portafolio["rendimiento"] = (portafolio["ganancia"] / (portafolio["cantidad"] * portafolio["precio_compra"])) * 100

en_perdida = portafolio[portafolio["rendimiento"] < 0]
portafolio.sort_values(by="rendimiento", ascending=False)
````

### 1. Construir un DataFrame con las columnas `ticker`, `cantidad`, `precio_compra`, `precio_actual`.

Para iniciar, transformaremos los datos de `mi_portafolio_dict` y `mis_precios_dict` a un formato adecuado para un DataFrame. Primero, los unificaremos en una lista de diccionarios, donde cada diccionario corresponderá a una fila en nuestro DataFrame.

````python
import pandas as pd

# Prepara los datos para el DataFrame
# Iteramos sobre los ítems del diccionario de portafolio para construir una lista de diccionarios.
# Cada diccionario contendrá 'ticker', 'cantidad', 'precio_compra' y 'precio_actual'.

datos_portafolio = []
for ticker, detalles in mi_portafolio_dict.items():
    datos_portafolio.append({
        "ticker": ticker,
        "cantidad": detalles["cantidad"],
        "precio_compra": detalles["precio_compra"],
        "precio_actual": mis_precios_dict[ticker]
    })

# Crea el DataFrame
portafolio_df = pd.DataFrame(datos_portafolio)
display(portafolio_df.round(2))
````

### 2. Agregar las columnas `valor_actual`, `ganancia` y `rendimiento`.

Ahora, procederemos a calcular las nuevas columnas. Para ello, utilizaremos las operaciones vectorizadas de pandas, que destacan por su eficiencia y eliminan la necesidad de emplear bucles `for` explícitos.

````python
# Calcula el valor actual de cada posición
portafolio_df["valor_actual"] = portafolio_df["cantidad"] * portafolio_df["precio_actual"]

# Calcula la ganancia (o pérdida) de cada posición
portafolio_df["ganancia"] = portafolio_df["valor_actual"] - (portafolio_df["cantidad"] * portafolio_df["precio_compra"])

# Calcula el rendimiento porcentual
# Asegúrate de no dividir por cero si el precio_compra * cantidad podría ser 0
portafolio_df["rendimiento"] = (portafolio_df["ganancia"] / (portafolio_df["cantidad"] * portafolio_df["precio_compra"])) * 100

display(portafolio_df.round(2))
````

### 3. Filtrar los activos en pérdida.

Para identificar los activos con rendimiento negativo, aplicaremos una condición de filtrado al DataFrame.

````python
# Filtra el DataFrame para mostrar solo los activos con rendimiento negativo
activos_en_perdida = portafolio_df[portafolio_df["rendimiento"] < 0]
display(activos_en_perdida.round(2))
````

### 4. Ordenar el DataFrame de mayor a menor rendimiento.

Finalmente, ordenaremos el portafolio completo en función de la columna `rendimiento`, de forma descendente.

````python
# Ordena el DataFrame completo por la columna 'rendimiento' de mayor a menor
portafolio_ordenado = portafolio_df.sort_values(by="rendimiento", ascending=False)
display(portafolio_ordenado.round(2))
````
